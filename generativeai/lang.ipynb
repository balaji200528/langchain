LLM MODEL INTEGRATION WITH THE GOOGLE GEMINI

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite"
)
```

```python
response=model.invoke("write an essay on the topic dogs are friendly to humans")
print(response.content)
```

streaming and batch

```python
response=model.invoke("write me code in ml to predict the price of a house given its size in square feet and number of bedrooms")
print(response.content)
```

```python
for chunk in model.stream("explain quantum computing in simple terms"):
    print(chunk.content,end="|",flush=True)
```

```python
r=model.batch(
    [
        "Explain the theory of relativity in simple terms.",
        "What are the benefits of using renewable energy sources?",
        "Describe the process of photosynthesis."
    ]
    config={
        'max_concurrency':2
    }
)


for response in r:
    response.content
```
